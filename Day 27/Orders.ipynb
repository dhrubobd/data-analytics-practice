```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pymysql
```

MySQL connection to Python

```python
sql_connection = pymysql.connect(
    host= "localhost",
    user= "root",
    password= "1234",
    database= "mydb"
)
```

```python
query = """select * from orders"""
pd.read_sql(query,sql_connection)
```

```python
query = """select * from orders where city='Dhaka'"""
pd.read_sql(query,sql_connection)
```

```python
query = """select * from orders where Amount>=50000"""
pd.read_sql(query,sql_connection)
```

```python
query = """
select City,
sum(Amount) as Revenue
from orders
group by City
order by Revenue desc
"""
pd.read_sql(query,sql_connection)
```
